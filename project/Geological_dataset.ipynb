GEOLOGICAL DATASET

```python
import pandas as pd
import os
```

Step 1: Load the original geolocation dataset

```python
input_path = r"D:\ITP\ECOMMERCE\Main Data\Cleaned_master_Olist_Data.csv"
df = pd.read_csv(input_path)
```

Step 2: Extract customer location columns

```python
customer_geo = df[
    [
        "customer_zip_code_prefix",
        "customer_city",
        "customer_state"
    ]
].copy()
```

Step 3: Rename columns

```python
customer_geo.columns = [
    "geolocation_zip_code_prefix",
    "geolocation_city",
    "geolocation_state"
]
```

Step 4: Extract seller location columns if available

```python
seller_columns = [
    "seller_zip_code_prefix",
    "seller_city",
    "seller_state"
]

if all(col in df.columns for col in seller_columns):
    seller_geo = df[seller_columns].copy()

    seller_geo.columns = [
        "geolocation_zip_code_prefix",
        "geolocation_city",
        "geolocation_state"
    ]

    # Combine customer and seller locations
    geo_df = pd.concat(
        [customer_geo, seller_geo],
        ignore_index=True
    )
else:
    geo_df = customer_geo.copy()
```

Step 5: Remove duplicate locations

```python
geo_df = geo_df.drop_duplicates().reset_index(drop=True)
```

create folder

```python
output_path = os.path.join(
    "D:\ITP\ECOMMERCE\Created dataset",
    "olist_geolocation_dataset.csv"
)

geo_df.to_csv(output_path, index=False)
```
