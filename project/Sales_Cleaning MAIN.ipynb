# Check Null Values

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

```python
# Null values
df["customer_city"].isna().sum()

# Blank values
df["customer_city"].str.strip().eq("").sum()

# Total missing values (null + blank)
df["customer_city"].isna().sum() + df["customer_city"].str.strip().eq("").sum()
```

```python
df["order_id"].duplicated().sum()
```

```python
df["order_status"].value_counts()
```

```python
df = df[df["order_status"]=="delivered"]
```

```python
df["order_status"].value_counts()
```

```python
df[[
    "product_category_name",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]].isnull().sum()
```

```python
df["product_category_name"] = df[
    "product_category_name"
].fillna("Unknown Category")
```

```python
median_weight = df["product_weight_g"].median()

df["product_weight_g"] = df[
    "product_weight_g"
].fillna(median_weight)
```

```python
median_length = df["product_length_cm"].median()

df["product_length_cm"] = df[
    "product_length_cm"
].fillna(median_length)
```

```python
median_height = df["product_height_cm"].median()

df["product_height_cm"] = df[
    "product_height_cm"
].fillna(median_height)
```

```python
median_width = df["product_width_cm"].median()

df["product_width_cm"] = df[
    "product_width_cm"
].fillna(median_width)
```

```python
df[[
    "payment_type",
    "payment_installments",
    "payment_value"
]].isnull().sum()
```

```python
df["payment_type"].value_counts()
```

```python
df["payment_type"].mode()
```

```python
df.groupby("payment_installments")[
    "payment_value"
].mean()
```

```python
df["payment_value"].isnull().sum()
```

```python
df = df.dropna(subset=["payment_value"])
```

```python
df["review_score"].isnull().sum()
```

```python
df = df.dropna(subset=["review_score"])
```

```python
df["review_score"].value_counts()
```

```python
df.duplicated().sum()
```

```python
df["seller_id"].duplicated().sum()
```

```python
df["seller_city"].str.strip().eq("").sum()
```

```python
df["seller_state"].str.strip().eq("").sum()
```

```python
df["seller_city"].fillna(
    "Unknown",
    inplace=True
)

df["seller_state"].fillna(
    "Unknown",
    inplace=True
)
```
