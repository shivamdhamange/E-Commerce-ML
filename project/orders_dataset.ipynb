```python
import pandas as pd
import os
```

```python

# Step 1: Load the cleaned dataset
input_path = r"D:\ITP\ECOMMERCE\Main Data\Cleaned_master_Olist_Data.csv"
df = pd.read_csv(input_path)
```

```python

# Step 2: Define the required order columns
order_columns = [
    "order_id",
    "customer_id",
    "order_status",
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]
```

```python

# Step 3: Check whether all required columns exist
missing_columns = [
    col for col in order_columns if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"These columns are missing from the cleaned dataset: {missing_columns}"
    )
```

```python

# Step 4: Extract the orders dataset
orders_df = df[order_columns].copy()
```

```python

# Step 5: Remove duplicate orders, if any
orders_df = orders_df.drop_duplicates(
    subset=["order_id"]
).reset_index(drop=True)


```

```python

# Step 6: Save the orders dataset
output_path = os.path.join(
    "D:\ITP\ECOMMERCE\Created dataset",
    "olist_orders_dataset.csv"
)

orders_df.to_csv(output_path, index=False)
```

```python

# Step 7: Verify the output
print("Orders dataset created successfully!")
print("File location:", output_path)
print("Dataset shape:", orders_df.shape)
print("\nColumns:")
print(orders_df.columns.tolist())

print("\nFirst 5 rows:")
print(orders_df.head())

print("\nDuplicate order IDs:",
      orders_df["order_id"].duplicated().sum())
```
