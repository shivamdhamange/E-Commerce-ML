```python
import pandas as pd
import os
```

```python

# Step 1: Load the cleaned dataset
input_path = r"C:\Users\Asus\OneDrive\Desktop\Falguni\Main data which we have to work\Cleaned_master_Olist_Data.csv"
df = pd.read_csv(input_path)
```

```python

# Step 2: Define the required columns
item_columns = [
    "order_id",
    "order_item_id",
    "product_id",
    "seller_id",
    "shipping_limit_date",
    "price",
    "freight_value"
]
```

```python

# Step 3: Check for missing columns
missing_columns = [
    col for col in item_columns if col not in df.columns
]

if missing_columns:
    raise ValueError(f"Missing columns: {missing_columns}")
```

```python

# Step 4: Extract order item information
order_items_df = df[item_columns].copy()

```

```python

# Step 5: Save the CSV file
output_path = os.path.join(
    "C:/Users/Asus/OneDrive/Desktop/Falguni/Created dataset",
    "olist_order_items_dataset.csv"
)

order_items_df.to_csv(output_path, index=False)
```

```python

# Step 6: Verify the output
print("Order items dataset created successfully!")
print("Saved at:", output_path)
print("Dataset shape:", order_items_df.shape)
print("\nFirst 5 rows:")
print(order_items_df.head())
```
