```python
import pandas as pd
import os
```

```python

# Step 1: Load the cleaned dataset
input_path = r"D:\ITP\ECOMMERCE\Main Data\Cleaned_master_Olist_Data.csv"
df = pd.read_csv(input_path)
```

```python

# Step 2: Define the product columns
product_columns = [
    "product_id",
    "product_category_name",
    "product_name_lenght",
    "product_description_lenght",
    "product_photos_qty",
    "product_weight_g",
    "product_length_cm",
    "product_height_cm",
    "product_width_cm"
]
```

```python

# Step 3: Check which product columns exist
available_columns = [
    col for col in product_columns if col in df.columns
]

if "product_id" not in available_columns:
    raise ValueError(
        "product_id was not found in the cleaned dataset. "
        "Check your dataset columns."
    )
```

```python

# Step 4: Extract available product information
products_df = df[available_columns].copy()
```

```python

# Step 5: Keep one row per product
products_df = products_df.drop_duplicates(
    subset=["product_id"]
).reset_index(drop=True)
```

```python

# Step 6: Save the products dataset
output_path = os.path.join(
    "D:\ITP\ECOMMERCE\Created dataset",
    "olist_products_dataset.csv"
)

products_df.to_csv(output_path, index=False)
```

```python

# Step 7: Verify the output
print("Products dataset created successfully!")
print("Saved at:", output_path)
print("Dataset shape:", products_df.shape)
print("Columns:", products_df.columns.tolist())
print(products_df.head())
```
