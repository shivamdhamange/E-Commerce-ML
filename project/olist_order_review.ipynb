```python
import pandas as pd
import os
```

# Step 1: Load the cleaned orders dataset

```python
orders_df = pd.read_csv(
    r"D:\ITP\ECOMMERCE\Main Data\Cleaned_master_Olist_Data.csv"
)
```

# Step 2: Create a reviews dataset using order IDs

```python
reviews_df = [
    "review_id",
    "order_id",
    "review_score",
    "review_comment_title",
    "review_comment_message",
    "review_creation_date",
    "review_answer_timestamp"
]

review_df = orders_df[reviews_df].copy()
```

# Step 4: Save the reviews dataset

```python

output_path = os.path.join(
    "D:\ITP\ECOMMERCE\Created dataset",
    "olist_order_reviews_dataset.csv"
)

review_df.to_csv(output_path, index=False)
```


# Step 5: Verify the output
