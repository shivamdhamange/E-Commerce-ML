SELLER DATASET

```python
import pandas as pd
import os
```

```python

# Step 1: Load the cleaned dataset
input_path = r"D:\ITP\ECOMMERCE\Main Data\Cleaned_master_Olist_Data.csv"
df = pd.read_csv(input_path)
```

```python

# Step 2: Select seller columns
seller_columns = [
    "seller_id",
    "seller_zip_code_prefix",
    "seller_city",
    "seller_state"
]
```

```python

# Step 3: Extract seller information
sellers_df = df[seller_columns].copy()
```

```python

# Step 4: Remove duplicate sellers
sellers_df = sellers_df.drop_duplicates(
    subset=["seller_id"]
).reset_index(drop=True)

```

```python

# Step 6: Save sellers dataset as CSV
output_path = os.path.join(
    "D:\ITP\ECOMMERCE\Created dataset",
    "olist_sellers_dataset.csv"
)

sellers_df.to_csv(output_path, index=False)
```
