```python
import pandas as pd
import os
```

```python

# 1. Load the cleaned Olist dataset
input_path = r"C:\Users\Asus\OneDrive\Desktop\Falguni\Main data which we have to work\Cleaned_master_Olist_Data.csv"

df = pd.read_csv(input_path)
```

```python

# 2. Select columns from the enhanced reference dataset
enhanced_columns = [
    "order_id",
    "order_item_id",
    "product_id",
    "seller_id",
    "shipping_limit_date",
    "price",
    "freight_value",
    "product_cost",
    "discount",
    "return_loss",
    "profit"
]
```

```python

# 3. Check that all required columns exist
missing_columns = [
    col for col in enhanced_columns
    if col not in df.columns
]

if missing_columns:
    raise ValueError(
        f"Missing columns in cleaned dataset: {missing_columns}"
    )
```

```python

# 4. Extract the required columns
enhanced_df = df[enhanced_columns].copy()
```

```python

# 5. Remove duplicate order-item records
enhanced_df = enhanced_df.drop_duplicates(
    subset=["order_id", "order_item_id"]
)
```

```python

# 6. Save the enhanced dataset
output_path = os.path.join(
    "C:/Users/Asus/OneDrive/Desktop/Falguni/Created dataset",
    "olist_order_items_enhanced.csv"
)

enhanced_df.to_csv(output_path, index=False)
```

```python

# 7. Display the results
print("Enhanced order items dataset created successfully!")
print("Saved at:", output_path)
print("Dataset shape:", enhanced_df.shape)
print("Total rows:", len(enhanced_df))
print("Total columns:", len(enhanced_df.columns))
print("\nColumns:")
print(enhanced_df.columns.tolist())
print("\nFirst five rows:")
print(enhanced_df.head())
```
