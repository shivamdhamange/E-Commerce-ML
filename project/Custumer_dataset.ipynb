CUSTOMER DATA EXTRACT FROM CLEANED DATASET

```python
import pandas as pd
import os
```

Step 1: Load your cleaned dataset

```python
df = pd.read_csv(r"C:\Users\Asus\OneDrive\Desktop\Falguni\Main data which we have to work\Cleaned_master_Olist_Data.csv")
```

Step 2: Select the customer columns

```python
customer_columns = [
    "customer_id",
    "customer_unique_id",
    "customer_zip_code_prefix",
    "customer_city",
    "customer_state"
]
```

Step 3: Extract customer data

```python
customers_df = df[customer_columns].drop_duplicates(
    subset=["customer_id"]
)
```

Step 4: Save as a separate CSV file

```python
output_path = os.path.join(
    "C:/Users/Asus/OneDrive/Desktop/Falguni/Created dataset",
    "olist_customers_dataset.csv"
)
```

```python
customers_df.to_csv(output_path, index=False)
```

Step 5: Verify the output

```python
print("Customer dataset created successfully!")
print("Shape:", customers_df.shape)
customers_df.head()
```
